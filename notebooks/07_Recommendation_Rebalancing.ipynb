```python
# AAPL MODEL — EXPECTED CALIBRATION ERROR (ECE)
# ============================================================

import numpy as np
import pandas as pd

calibration_check = pd.DataFrame({
    "actual": np.asarray(y_val).astype(int),
    "probability": np.asarray(val_prob_class_1)})

# 10 equal-width probability bins
calibration_check["bin"] = pd.cut(
    calibration_check["probability"],
    bins=np.linspace(0, 1, 11),
    include_lowest=True)

ece_rows = []

for bin_name, group in calibration_check.groupby("bin", observed=True):

    if len(group) == 0:
        continue

    predicted = group["probability"].mean()
    actual = group["actual"].mean()
    weight = len(group) / len(calibration_check)

    ece_rows.append({
        "bin": str(bin_name),
        "observations": len(group),
        "mean_predicted": predicted,
        "actual_positive_rate": actual,
        "absolute_calibration_error": abs(predicted - actual),
        "weighted_error": weight * abs(predicted - actual)})

ece_df = pd.DataFrame(ece_rows)

ece = ece_df["weighted_error"].sum()

print("=" * 70)
print("AAPL MODEL — EXPECTED CALIBRATION ERROR")
print("=" * 70)

print(f"\nECE: {ece:.6f}")

print("\nCalibration table:")
display(ece_df)

print("\nInterpretation:")
if ece < 0.05:
    print("Very low calibration error.")
elif ece < 0.10:
    print("Moderate calibration error.")
else:
    print("Meaningful calibration error — raw probabilities should be treated cautiously.")
```

```python
# AAPL — VALIDATION SIGNAL BUCKET ANALYSIS

import numpy as np
import pandas as pd
from pathlib import Path

validation_policy = pd.DataFrame({
    "actual": np.asarray(y_val).astype(int),
    "positive_probability": np.asarray(val_prob_class_1)})

validation_policy["non_positive_probability"] = (
    1 - validation_policy["positive_probability"])

validation_policy["probability_margin"] = (
    validation_policy["positive_probability"] - 0.50
).abs()

validation_policy["predicted_direction"] = np.where(
    validation_policy["positive_probability"] >= 0.50,
    "POSITIVE",
    "NON-POSITIVE")

validation_policy["signal_bucket"] = pd.cut(
    validation_policy["positive_probability"],
    bins=[0.0, 0.40, 0.45, 0.50, 0.55, 0.60, 1.0],
    labels=[
        "<=40%",
        "40–45%",
        "45–50%",
        "50–55%",
        "55–60%",
        ">60%"
    ],
    include_lowest=True)

bucket_summary = (
    validation_policy
    .groupby("signal_bucket", observed=True)
    .agg(
        observations=("actual", "size"),
        mean_predicted_probability=(
            "positive_probability", "mean"
        ),
        actual_positive_rate=("actual", "mean"),
        mean_probability_margin=(
            "probability_margin", "mean"))
    .reset_index())

bucket_summary["actual_positive_rate"] *= 100
bucket_summary["mean_predicted_probability"] *= 100
bucket_summary["mean_probability_margin"] *= 100


policy_validation_output = (
    PROJECT_ROOT
    / "results"
    / "model_comparison"
    / "aapl_validation_signal_buckets.csv")

bucket_summary.to_csv(
    policy_validation_output,
    index=False)

print("=" * 70)
print("AAPL — VALIDATION SIGNAL BUCKET ANALYSIS")
print("=" * 70)

display(bucket_summary)

print(f"\nValidation observations: {len(validation_policy)}")

print(f"\nSaved to:")
print(policy_validation_output)

print("\nStatus: Validation signal analysis completed.")
```

```python
# AAPL — VALIDATION THRESHOLD POLICY ANALYSIS

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score)

thresholds = np.arange(0.40, 0.61, 0.025)

threshold_results = []

for threshold in thresholds:

    predicted = (
        validation_policy["positive_probability"] >= threshold
    ).astype(int)

    positive_signal_rate = predicted.mean()

    threshold_results.append({
        "threshold": threshold,
        "accuracy": accuracy_score(
            validation_policy["actual"],
            predicted),

        "precision": precision_score(
            validation_policy["actual"],
            predicted,
            zero_division=0),

        "recall": recall_score(
            validation_policy["actual"],
            predicted,
            zero_division=0),

        "f1": f1_score(
            validation_policy["actual"],
            predicted,
            zero_division=0),

        "positive_signal_rate": positive_signal_rate})

threshold_results_df = pd.DataFrame(
    threshold_results)

# Convert rates to percentages
for column in [
    "accuracy",
    "precision",
    "recall",
    "f1",
    "positive_signal_rate"
]:
    threshold_results_df[column] *= 100

threshold_output = (
    PROJECT_ROOT
    / "results"
    / "model_comparison"
    / "aapl_validation_threshold_policy.csv")

threshold_results_df.to_csv(
    threshold_output,
    index=False)

print("=" * 70)
print("AAPL — VALIDATION THRESHOLD POLICY ANALYSIS")
print("=" * 70)

display(
    threshold_results_df.round(3))

print("\nValidation observations:", len(validation_policy))

print("\nSaved to:")
print(threshold_output)

print("\nStatus: Threshold policy analysis completed.")
```

```python
# AAPL — REAL-PRICE BACKTEST POLICY ANALYSIS

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

REAL_BACKTEST_PATH = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_real_price_backtest_predictions.csv")

if not REAL_BACKTEST_PATH.exists():
    raise FileNotFoundError(
        f"File not found:\n{REAL_BACKTEST_PATH}")

real_bt = pd.read_csv(REAL_BACKTEST_PATH)

real_bt["date"] = pd.to_datetime(real_bt["date"])

real_bt = (
    real_bt
    .sort_values("date")
    .reset_index(drop=True))


required_columns = [
    "date",
    "actual_5day_return",
    "target",
    "prediction_probability",
    "signal",
    "net_strategy_return"]

missing_columns = [
    col for col in required_columns
    if col not in real_bt.columns]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}")

real_bt["probability_bucket"] = pd.cut(
    real_bt["prediction_probability"],
    bins=[0.0, 0.40, 0.45, 0.50, 0.55, 0.60, 0.70, 1.0],
    labels=[
        "<=40%",
        "40–45%",
        "45–50%",
        "50–55%",
        "55–60%",
        "60–70%",
        ">70%"],
    include_lowest=True)

bucket_financial_summary = (
    real_bt
    .groupby("probability_bucket", observed=True)
    .agg(
        observations=("actual_5day_return", "size"),
        mean_predicted_probability=(
            "prediction_probability",
            "mean"),
        actual_positive_rate=(
            "target",
            "mean"),
        mean_actual_5day_return=(
            "actual_5day_return",
            "mean"),
        median_actual_5day_return=(
            "actual_5day_return",
            "median"),
        mean_net_strategy_return=(
            "net_strategy_return",
            "mean"))
    .reset_index())

# Percentage conversion
for col in [
    "mean_predicted_probability",
    "actual_positive_rate",
    "mean_actual_5day_return",
    "median_actual_5day_return",
    "mean_net_strategy_return"]:
    bucket_financial_summary[col] *= 100

financial_output = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_probability_financial_outcomes.csv")

bucket_financial_summary.to_csv(
    financial_output,
    index=False)

print("=" * 80)
print("AAPL — REAL-PRICE BACKTEST PROBABILITY ANALYSIS")
print("=" * 80)

print(
    f"\nObservations: {len(real_bt)}")

print(
    f"Date range: "
    f"{real_bt['date'].min().date()} → "
    f"{real_bt['date'].max().date()}")

display(
    bucket_financial_summary.round(4))

print(f"\nSaved to:")
print(financial_output)

print("\nStatus: Financial probability analysis completed.")
```

```python
# AAPL — SIGNAL POLICY AUDIT
# Empirical threshold analysis for recommendation engine

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

REAL_BACKTEST_PATH = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_real_price_backtest_predictions.csv")

df = pd.read_csv(REAL_BACKTEST_PATH)
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)

# Evaluate probability thresholds

thresholds = np.arange(0.40, 0.71, 0.05)

rows = []

for threshold in thresholds:

    signal = (
        df["prediction_probability"] >= threshold
    ).astype(int)

    selected = df[signal == 1].copy()

    if len(selected) == 0:
        continue

    hit_rate = selected["target"].mean()
    avg_return = selected["actual_5day_return"].mean()
    median_return = selected["actual_5day_return"].median()
    avg_net_return = selected["net_strategy_return"].mean()

    cumulative_return = (
        (1 + selected["net_strategy_return"]).prod() - 1)

    rows.append({
        "threshold": threshold,
        "observations": len(selected),
        "coverage_pct": len(selected) / len(df) * 100,
        "positive_hit_rate_pct": hit_rate * 100,
        "mean_actual_5d_return_pct": avg_return * 100,
        "median_actual_5d_return_pct": median_return * 100,
        "mean_net_strategy_return_pct": avg_net_return * 100,
        "cumulative_selected_return_pct": cumulative_return * 100})

policy_audit = pd.DataFrame(rows)

output_path = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_signal_policy_audit.csv")

policy_audit.to_csv(output_path, index=False)

print("=" * 80)
print("AAPL — SIGNAL POLICY AUDIT")
print("=" * 80)

display(policy_audit.round(4))

print("\nSaved to:")
print(output_path)

print("\nStatus: Signal policy audit completed.")
```

```python
# AAPL — RISK-AWARE SIGNAL POLICY AUDIT

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

REAL_BACKTEST_PATH = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_real_price_backtest_predictions.csv")

df = pd.read_csv(REAL_BACKTEST_PATH)

df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)

thresholds = np.arange(0.40, 0.71, 0.05)

rows = []

for threshold in thresholds:

    selected = df[
        df["prediction_probability"] >= threshold
    ].copy()

    if len(selected) < 2:
        continue

    returns = selected["net_strategy_return"].astype(float)

    mean_return = returns.mean()
    std_return = returns.std(ddof=1)

    # Annualized volatility approximation
    annualized_vol = std_return * np.sqrt(252)

    # Sharpe-like statistic using zero risk-free rate
    if std_return > 0:
        sharpe = (
            mean_return / std_return
        ) * np.sqrt(252)
    else:
        sharpe = np.nan

    # Sequential equity curve for selected observations
    equity = (1 + returns).cumprod()

    running_peak = equity.cummax()

    drawdown = (
        equity / running_peak - 1)

    max_drawdown = drawdown.min()

    rows.append({
        "threshold": threshold,
        "observations": len(selected),
        "coverage_pct": len(selected) / len(df) * 100,
        "positive_hit_rate_pct": (
            selected["target"].mean() * 100
        ),
        "mean_5d_return_pct": (
            selected["actual_5day_return"].mean() * 100
        ),
        "mean_net_return_pct": (
            mean_return * 100
        ),
        "annualized_volatility_pct": (
            annualized_vol * 100
        ),
        "sharpe_like": sharpe,
        "max_drawdown_pct": (
            max_drawdown * 100)})

risk_policy_audit = pd.DataFrame(rows)

output_path = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_risk_aware_signal_policy_audit.csv")

risk_policy_audit.to_csv(
    output_path,
    index=False)

print("=" * 80)
print("AAPL — RISK-AWARE SIGNAL POLICY AUDIT")
print("=" * 80)

display(
    risk_policy_audit.round(4))

print("\nSaved to:")
print(output_path)

print("\nStatus: Risk-aware signal policy audit completed.")
```

```python
# AAPL — CHRONOLOGICAL SIGNAL POLICY VALIDATION
# Train / Validation / Test policy audit

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

REAL_BACKTEST_PATH = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_real_price_backtest_predictions.csv")

df = pd.read_csv(REAL_BACKTEST_PATH)

df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)

# Chronological split

n = len(df)

train_end = int(n * 0.70)
validation_end = int(n * 0.85)

df["period"] = "Test"

df.loc[:train_end - 1, "period"] = "Train"
df.loc[train_end:validation_end - 1, "period"] = "Validation"

print("=" * 80)
print("AAPL — CHRONOLOGICAL SIGNAL POLICY VALIDATION")
print("=" * 80)

print("\nPeriod ranges:")

for period in ["Train", "Validation", "Test"]:
    part = df[df["period"] == period]

    print(
        f"{period:12s}: "
        f"{part['date'].min().date()} → "
        f"{part['date'].max().date()} "
        f"({len(part)} observations)")

# Threshold evaluation

thresholds = np.arange(0.40, 0.71, 0.05)

rows = []

for period in ["Train", "Validation", "Test"]:

    part = df[df["period"] == period].copy()

    for threshold in thresholds:

        selected = part[
            part["prediction_probability"] >= threshold
        ].copy()

        if len(selected) == 0:
            continue

        returns = selected[
            "net_strategy_return"
        ].astype(float)

        std_return = returns.std(ddof=1)

        if std_return > 0:
            sharpe_like = (
                returns.mean()
                / std_return
                * np.sqrt(252))
        else:
            sharpe_like = np.nan

        equity = (1 + returns).cumprod()

        running_peak = equity.cummax()

        drawdown = (
            equity / running_peak - 1)

        rows.append({
            "period": period,
            "threshold": threshold,
            "observations": len(selected),
            "coverage_pct": (
                len(selected) / len(part) * 100
            ),
            "positive_hit_rate_pct": (
                selected["target"].mean() * 100
            ),
            "mean_actual_5d_return_pct": (
                selected["actual_5day_return"].mean()
                * 100
            ),
            "mean_net_return_pct": (
                returns.mean() * 100
            ),
            "sharpe_like": sharpe_like,
            "max_drawdown_pct": (
                drawdown.min() * 100
            )})

policy_oos = pd.DataFrame(rows)

output_path = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_chronological_signal_policy_validation.csv")

policy_oos.to_csv(
    output_path,
    index=False)

display(
    policy_oos.round(4))

print("\nSaved to:")
print(output_path)

print("\nStatus: Chronological policy validation completed.")
```

```python
# AAPL — VALIDATION-ONLY POLICY SELECTION
# Independent signal reconstruction

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

PATH = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_real_price_backtest_predictions.csv")

df = pd.read_csv(PATH)

df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)

# Same chronological split used in previous audit

n = len(df)

train_end = int(n * 0.70)
validation_end = int(n * 0.85)

validation = df.iloc[
    train_end:validation_end
].copy()

TRANSACTION_COST = 0.001

thresholds = [0.50, 0.55, 0.60, 0.65]

rows = []

for threshold in thresholds:

    signal = (
        validation["prediction_probability"]
        >= threshold
    ).astype(int)

    # Independent strategy return reconstruction
    gross_return = (
        signal
        * validation["actual_5day_return"])

    net_return = (
        gross_return
        - signal * TRANSACTION_COST)

    active = signal == 1

    selected_returns = (
        validation.loc[active, "actual_5day_return"])

    selected_net = net_return[active]

    if len(selected_net) > 1:
        std = selected_net.std(ddof=1)

        sharpe = (
            selected_net.mean()
            / std
            * np.sqrt(252)
            if std > 0
            else np.nan)
    else:
        sharpe = np.nan

    rows.append({
        "threshold": threshold,
        "active_observations": int(active.sum()),
        "coverage_pct": (
            active.mean() * 100),
        "positive_hit_rate_pct": (
            validation.loc[active, "target"].mean()
            * 100
            if active.sum() > 0
            else np.nan),
        "mean_actual_5d_return_pct": (
            selected_returns.mean() * 100
            if len(selected_returns) > 0
            else np.nan),
        "mean_net_return_pct": (
            selected_net.mean() * 100
            if len(selected_net) > 0
            else np.nan),
        "total_net_return_pct": (
            selected_net.sum() * 100
            if len(selected_net) > 0
            else np.nan),
        "sharpe_like": sharpe})

validation_policy = pd.DataFrame(rows)

output_path = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_validation_policy_selection.csv")

validation_policy.to_csv(
    output_path,
    index=False)

print("=" * 80)
print("AAPL — VALIDATION-ONLY POLICY SELECTION")
print("=" * 80)

print(
    f"\nValidation period: "
    f"{validation['date'].min().date()} → "
    f"{validation['date'].max().date()}")

display(
    validation_policy.round(4))

print("\nSaved to:")
print(output_path)

print(
    "\nImportant: Test period was NOT used for policy selection.")
```

```python
# AAPL — FROZEN POLICY FINAL TEST AUDIT
# Policy frozen from validation: probability >= 60%
# Test period remains completely untouched for selection

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

PATH = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_real_price_backtest_predictions.csv")

df = pd.read_csv(PATH)

df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").reset_index(drop=True)

# Chronological split

n = len(df)

train_end = int(n * 0.70)
validation_end = int(n * 0.85)

test = df.iloc[validation_end:].copy()

# FROZEN POLICY

FROZEN_THRESHOLD = 0.60
TRANSACTION_COST = 0.001

test["policy_signal"] = (
    test["prediction_probability"]
    >= FROZEN_THRESHOLD
).astype(int)

# Independent return reconstruction
test["gross_policy_return"] = (
    test["policy_signal"]
    * test["actual_5day_return"])

test["net_policy_return"] = (
    test["gross_policy_return"]
    - test["policy_signal"] * TRANSACTION_COST)

# Metrics

active = test["policy_signal"] == 1

active_returns = test.loc[
    active,
    "actual_5day_return"]

active_net_returns = test.loc[
    active,
    "net_policy_return"]

hit_rate = (
    test.loc[active, "target"].mean()
    if active.sum() > 0
    else np.nan)

mean_actual = (
    active_returns.mean()
    if len(active_returns) > 0
    else np.nan)

mean_net = (
    active_net_returns.mean()
    if len(active_net_returns) > 0
    else np.nan)

total_net = (
    active_net_returns.sum()
    if len(active_net_returns) > 0
    else np.nan)

std_net = (
    active_net_returns.std(ddof=1)
    if len(active_net_returns) > 1
    else np.nan)

sharpe_like = (
    mean_net / std_net * np.sqrt(252)
    if pd.notna(std_net) and std_net > 0
    else np.nan)

# Sequential equity curve

test["policy_equity"] = (
    1 + test["net_policy_return"]
).cumprod()

test["buy_hold_equity"] = (
    1 + test["actual_5day_return"]
).cumprod()

policy_peak = test["policy_equity"].cummax()

test["policy_drawdown"] = (
    test["policy_equity"] / policy_peak - 1)

max_drawdown = test["policy_drawdown"].min()

# Summary

summary = pd.DataFrame([{
    "policy_threshold": FROZEN_THRESHOLD,
    "test_start": test["date"].min().date(),
    "test_end": test["date"].max().date(),
    "test_observations": len(test),
    "active_observations": int(active.sum()),
    "coverage_pct": active.mean() * 100,
    "positive_hit_rate_pct": hit_rate * 100,
    "mean_actual_5d_return_pct": mean_actual * 100,
    "mean_net_return_pct": mean_net * 100,
    "total_net_return_sum_pct": total_net * 100,
    "sharpe_like": sharpe_like,
    "max_drawdown_pct": max_drawdown * 100}])

output_path = (
    PROJECT_ROOT
    / "results"
    / "real_price_backtest"
    / "aapl_frozen_60pct_policy_test_audit.csv")

summary.to_csv(
    output_path,
    index=False)

print("=" * 80)
print("AAPL — FROZEN 60% POLICY FINAL TEST AUDIT")
print("=" * 80)

display(summary.round(4))

print("\nSaved to:")
print(output_path)

print(
    "\nPolicy status: FROZEN FROM VALIDATION — TEST USED ONLY FOR FINAL AUDIT.")
```

```python
# PORTFOLIO — CREATE 5-DAY TARGETS
# ROBUST MULTIINDEX / FLAT-COLUMN VERSION

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

FEATURE_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "features.parquet")

OUTPUT_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed")

df = pd.read_parquet(FEATURE_PATH).copy()
df = df.sort_index()

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

print("=" * 80)
print("PORTFOLIO — 5-DAY TARGET CREATION")
print("=" * 80)

# Helper: locate Adj Close column

def find_adj_close_column(dataframe, ticker):

    # MultiIndex case
    if isinstance(dataframe.columns, pd.MultiIndex):

        for col in dataframe.columns:

            parts = [
                str(x).strip().lower()
                for x in col]

            if (
                "adj close" in parts
                and ticker.lower() in parts):
                return col

    # Flat-column fallback
    else:

        candidates = [
            col
            for col in dataframe.columns
            if (
                "adj close" in str(col).lower()
                and ticker.lower() in str(col).lower())]

        if candidates:
            return candidates[0]

    return None


target_summary = []

for ticker in UNIVERSE:

    price_col = find_adj_close_column(
        df,
        ticker)

    if price_col is None:
        raise ValueError(
            f"Could not locate Adj Close for {ticker}")

    future_return_col = (
        f"future_return_5d_{ticker}")

    target_col = (
        f"target_5d_{ticker}")

    price = pd.to_numeric(
        df[price_col],
        errors="coerce")

    # Future 5-day return

    df[future_return_col] = (
        price.shift(-5)
        / price
        - 1)

    # Binary direction target

    df[target_col] = np.where(
        df[future_return_col].notna(),
        (
            df[future_return_col] > 0
        ).astype(float),
        np.nan)

    valid = df[future_return_col].notna()

    target_summary.append({
        "ticker": ticker,
        "price_column": str(price_col),
        "valid_observations": int(valid.sum()),
        "positive_observations": int(
            (
                df.loc[valid, target_col] == 1
            ).sum()),
        "negative_or_zero_observations": int(
            (
                df.loc[valid, target_col] == 0
            ).sum()),
        "positive_rate_pct": (
            df.loc[valid, target_col].mean()
            * 100)})

target_summary = pd.DataFrame(
    target_summary)

output_path = (
    OUTPUT_DIR
    / "portfolio_features_with_5d_targets.parquet")

df.to_parquet(
    output_path)

summary_path = (
    PROJECT_ROOT
    / "results"
    / "model_comparison"
    / "portfolio_5d_target_summary.csv")

summary_path.parent.mkdir(
    parents=True,
    exist_ok=True)

target_summary.to_csv(
    summary_path,
    index=False)

print("\nTarget summary:")
display(
    target_summary.round(4))

print(
    "\nOutput dataset shape:",
    df.shape)

print("\nSaved dataset:")
print(output_path)

print("\nSaved target summary:")
print(summary_path)

print(
    "\nStatus: 5-day targets created successfully "
    "for all 10 stocks.")
```

```python
# PORTFOLIO — 10-STOCK LOGISTIC REGRESSION BASELINE
# Leakage-safe chronological training

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score)

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

INPUT_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "portfolio_features_with_5d_targets.parquet")

df = pd.read_parquet(INPUT_PATH).copy()
df = df.sort_index()

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

# Leakage-safe common feature set

feature_columns = [
    col
    for col in df.columns
    if not (
        "future_return_5d_" in str(col)
        or "target_5d_" in str(col))]

X = df[feature_columns].select_dtypes(
    include=[np.number]
).copy()

# Remove warm-up rows

complete_mask = ~X.isna().any(axis=1)

model_df = df.loc[complete_mask].copy()
X = X.loc[complete_mask]

# Chronological split

n = len(model_df)

train_end = int(n * 0.70)
validation_end = int(n * 0.85)

X_train = X.iloc[:train_end]
X_val = X.iloc[train_end:validation_end]
X_test = X.iloc[validation_end:]

results = []

print("=" * 80)
print("PORTFOLIO — LOGISTIC REGRESSION BASELINE")
print("=" * 80)

print(
    f"\nTrain: {X_train.index.min().date()} "
    f"→ {X_train.index.max().date()}")

print(
    f"Validation: {X_val.index.min().date()} "
    f"→ {X_val.index.max().date()}")

print(
    f"Test: {X_test.index.min().date()} "
    f"→ {X_test.index.max().date()}")

# Train one model per stock

for ticker in UNIVERSE:

    target_col = f"target_5d_{ticker}"

    y = df.loc[
        model_df.index,
        target_col]

    y_train = y.iloc[:train_end]
    y_val = y.iloc[train_end:validation_end]
    y_test = y.iloc[validation_end:]

    # Remove target-less rows from each split
    train_mask = y_train.notna()
    val_mask = y_val.notna()
    test_mask = y_test.notna()

    Xtr = X_train.loc[train_mask]
    Xv = X_val.loc[val_mask]
    Xt = X_test.loc[test_mask]

    ytr = y_train.loc[train_mask].astype(int)
    yv = y_val.loc[val_mask].astype(int)
    yt = y_test.loc[test_mask].astype(int)

    # Leakage-safe preprocessing

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            LogisticRegression(
                max_iter=2000,
                random_state=42
            ))
    ])

    model.fit(Xtr, ytr)

    # Validation predictions

    val_pred = model.predict(Xv)
    val_prob = model.predict_proba(Xv)[:, 1]

    # Test predictions

    test_pred = model.predict(Xt)
    test_prob = model.predict_proba(Xt)[:, 1]

    # Metrics

    results.append({
        "ticker": ticker,

        "train_rows": len(ytr),
        "validation_rows": len(yv),
        "test_rows": len(yt),

        "validation_accuracy": accuracy_score(
            yv, val_pred),
        "validation_precision": precision_score(
            yv, val_pred, zero_division=0),
        "validation_recall": recall_score(
            yv, val_pred, zero_division=0),
        "validation_f1": f1_score(
            yv, val_pred, zero_division=0),
        "validation_auc": roc_auc_score(
            yv, val_prob),

        "test_accuracy": accuracy_score(
            yt, test_pred),
        "test_precision": precision_score(
            yt, test_pred, zero_division=0),
        "test_recall": recall_score(
            yt, test_pred, zero_division=0),
        "test_f1": f1_score(
            yt, test_pred, zero_division=0),
        "test_auc": roc_auc_score(
            yt, test_prob)})

results_df = pd.DataFrame(results)

output_path = (
    PROJECT_ROOT
    / "results"
    / "classical_ml"
    / "portfolio_logistic_baseline.csv")

output_path.parent.mkdir(
    parents=True,
    exist_ok=True)

results_df.to_csv(
    output_path,
    index=False)

print("\nResults:")
display(
    results_df.round(4))

print("\nSaved to:")
print(output_path)

print(
    "\nStatus: 10-stock Logistic Regression baseline completed.")
```

```python
# PORTFOLIO — 10-STOCK RANDOM FOREST BASELINE
# Leakage-safe chronological training

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score)

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

INPUT_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "portfolio_features_with_5d_targets.parquet")

df = pd.read_parquet(INPUT_PATH).copy()
df = df.sort_index()

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

# Leakage-safe features

feature_columns = [
    col
    for col in df.columns
    if not (
        "future_return_5d_" in str(col)
        or "target_5d_" in str(col)
    )
]

X = df[feature_columns].select_dtypes(
    include=[np.number]
).copy()

# Remove warm-up rows
complete_mask = ~X.isna().any(axis=1)

model_df = df.loc[complete_mask].copy()
X = X.loc[complete_mask]

# Chronological split

n = len(model_df)

train_end = int(n * 0.70)
validation_end = int(n * 0.85)

X_train = X.iloc[:train_end]
X_val = X.iloc[train_end:validation_end]
X_test = X.iloc[validation_end:]

results = []

print("=" * 80)
print("PORTFOLIO — RANDOM FOREST")
print("=" * 80)

# One model per stock

for ticker in UNIVERSE:

    target_col = f"target_5d_{ticker}"

    y = df.loc[
        model_df.index,
        target_col]

    y_train = y.iloc[:train_end]
    y_val = y.iloc[train_end:validation_end]
    y_test = y.iloc[validation_end:]

    train_mask = y_train.notna()
    val_mask = y_val.notna()
    test_mask = y_test.notna()

    Xtr = X_train.loc[train_mask]
    Xv = X_val.loc[val_mask]
    Xt = X_test.loc[test_mask]

    ytr = y_train.loc[train_mask].astype(int)
    yv = y_val.loc[val_mask].astype(int)
    yt = y_test.loc[test_mask].astype(int)

    # Model

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")),
        (
            "model",
            RandomForestClassifier(
                n_estimators=300,
                max_depth=8,
                min_samples_leaf=10,
                class_weight="balanced",
                n_jobs=-1,
                random_state=42))])

    model.fit(Xtr, ytr)

    # Predictions

    val_pred = model.predict(Xv)
    val_prob = model.predict_proba(Xv)[:, 1]

    test_pred = model.predict(Xt)
    test_prob = model.predict_proba(Xt)[:, 1]

    # Metrics

    results.append({
        "ticker": ticker,

        "train_rows": len(ytr),
        "validation_rows": len(yv),
        "test_rows": len(yt),

        "validation_accuracy": accuracy_score(
            yv, val_pred
        ),
        "validation_precision": precision_score(
            yv, val_pred, zero_division=0
        ),
        "validation_recall": recall_score(
            yv, val_pred, zero_division=0
        ),
        "validation_f1": f1_score(
            yv, val_pred, zero_division=0
        ),
        "validation_auc": roc_auc_score(
            yv, val_prob
        ),

        "test_accuracy": accuracy_score(
            yt, test_pred
        ),
        "test_precision": precision_score(
            yt, test_pred, zero_division=0
        ),
        "test_recall": recall_score(
            yt, test_pred, zero_division=0
        ),
        "test_f1": f1_score(
            yt, test_pred, zero_division=0
        ),
        "test_auc": roc_auc_score(
            yt, test_prob
        )})

results_df = pd.DataFrame(results)

output_path = (
    PROJECT_ROOT
    / "results"
    / "classical_ml"
    / "portfolio_random_forest.csv")

output_path.parent.mkdir(
    parents=True,
    exist_ok=True)

results_df.to_csv(
    output_path,
    index=False)

print("\nResults:")
display(
    results_df.round(4))

print("\nSaved to:")
print(output_path)

print(
    "\nStatus: 10-stock Random Forest baseline completed.")
```

```python
# PORTFOLIO — 10-STOCK SVR REGRESSION
# Predict 5-day future return
# Direction calculated from predicted return

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score)

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

INPUT_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "portfolio_features_with_5d_targets.parquet")

df = pd.read_parquet(INPUT_PATH).copy()
df = df.sort_index()

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

# Leakage-safe features

feature_columns = [
    col
    for col in df.columns
    if not (
        "future_return_5d_" in str(col)
        or "target_5d_" in str(col))]

X = df[feature_columns].select_dtypes(
    include=[np.number]
).copy()

# Remove feature warm-up rows

complete_mask = ~X.isna().any(axis=1)

model_df = df.loc[complete_mask].copy()
X = X.loc[complete_mask]

# Chronological split

n = len(model_df)

train_end = int(n * 0.70)
validation_end = int(n * 0.85)

X_train = X.iloc[:train_end]
X_val = X.iloc[train_end:validation_end]
X_test = X.iloc[validation_end:]

results = []

print("=" * 80)
print("PORTFOLIO — SVR REGRESSION")
print("=" * 80)

# Train one SVR model per stock

for ticker in UNIVERSE:

    future_col = f"future_return_5d_{ticker}"

    y = df.loc[
        model_df.index,
        future_col]

    y_train = y.iloc[:train_end]
    y_val = y.iloc[train_end:validation_end]
    y_test = y.iloc[validation_end:]

    train_mask = y_train.notna()
    val_mask = y_val.notna()
    test_mask = y_test.notna()

    Xtr = X_train.loc[train_mask]
    Xv = X_val.loc[val_mask]
    Xt = X_test.loc[test_mask]

    ytr = y_train.loc[train_mask].astype(float)
    yv = y_val.loc[val_mask].astype(float)
    yt = y_test.loc[test_mask].astype(float)

    # Leakage-safe preprocessing + SVR

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            SVR(
                C=0.1,
                gamma="scale",
                epsilon=0.01,
                kernel="rbf"))])

    print(f"\nTraining {ticker}...")

    model.fit(Xtr, ytr)

    # Predictions

    val_pred = model.predict(Xv)
    test_pred = model.predict(Xt)

    # Metrics helper

    def regression_metrics(
        y_true,
        y_pred
    ):

        rmse = np.sqrt(
            mean_squared_error(
                y_true,
                y_pred))

        mae = mean_absolute_error(
            y_true,
            y_pred)

        r2 = r2_score(
            y_true,
            y_pred)

        directional_accuracy = (
            np.sign(y_true)
            == np.sign(y_pred)
        ).mean() * 100

        return (
            rmse,
            mae,
            r2,
            directional_accuracy)

    (
        val_rmse,
        val_mae,
        val_r2,
        val_direction
    ) = regression_metrics(
        yv,
        val_pred)

    (
        test_rmse,
        test_mae,
        test_r2,
        test_direction
    ) = regression_metrics(
        yt,
        test_pred
    )

    results.append({
        "ticker": ticker,

        "train_rows": len(ytr),
        "validation_rows": len(yv),
        "test_rows": len(yt),

        "validation_RMSE": val_rmse,
        "validation_MAE": val_mae,
        "validation_R2": val_r2,
        "validation_directional_accuracy": val_direction,

        "test_RMSE": test_rmse,
        "test_MAE": test_mae,
        "test_R2": test_r2,
        "test_directional_accuracy": test_direction})

results_df = pd.DataFrame(results)

output_path = (
    PROJECT_ROOT
    / "results"
    / "classical_ml"
    / "portfolio_svr_regression.csv")

output_path.parent.mkdir(
    parents=True,
    exist_ok=True)

results_df.to_csv(
    output_path,
    index=False)

print("\n" + "=" * 80)
print("SVR RESULTS")
print("=" * 80)

display(
    results_df.round(6))

print("\nSaved to:")
print(output_path)

print(
    "\nStatus: 10-stock SVR regression completed.")
```

```python
# PORTFOLIO — SVR LATEST PREDICTION ENGINE
# Train on all available labeled data
# Generate latest 5-day return prediction for 10 stocks

import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

INPUT_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "portfolio_features_with_5d_targets.parquet")

df = pd.read_parquet(INPUT_PATH).copy()
df = df.sort_index()

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

# Leakage-safe feature selection

feature_columns = [
    col
    for col in df.columns
    if not (
        "future_return_5d_" in str(col)
        or "target_5d_" in str(col))]

X = df[feature_columns].select_dtypes(
    include=[np.number]
).copy()

# Keep only rows with complete features
complete_mask = ~X.isna().any(axis=1)

model_df = df.loc[complete_mask].copy()
X = X.loc[complete_mask]

# Latest feature row
latest_date = model_df.index.max()
X_latest = X.loc[[latest_date]]

results = []

print("=" * 80)
print("PORTFOLIO — SVR LATEST PREDICTION ENGINE")
print("=" * 80)

print(f"\nLatest feature date: {latest_date}")
print(f"Feature count: {X.shape[1]}")

# Train one SVR per stock using all available labeled rows

for ticker in UNIVERSE:

    future_col = f"future_return_5d_{ticker}"

    y = model_df[future_col]

    train_mask = y.notna()

    X_train = X.loc[train_mask]
    y_train = y.loc[train_mask].astype(float)

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            SVR(
                C=0.1,
                gamma="scale",
                epsilon=0.01,
                kernel="rbf"))])

    print(f"\nTraining {ticker}...")

    model.fit(
        X_train,
        y_train)

    predicted_return = float(
        model.predict(X_latest)[0])

    predicted_direction = (
        "POSITIVE"
        if predicted_return > 0
        else "NON-POSITIVE")

    results.append({
        "date": latest_date,
        "ticker": ticker,
        "model": "SVR",
        "predicted_5d_return": predicted_return,
        "predicted_5d_return_pct": predicted_return * 100,
        "predicted_direction": predicted_direction,
        "training_rows": len(y_train)})

# Results

predictions_df = pd.DataFrame(results)

predictions_df = predictions_df.sort_values(
    "predicted_5d_return",
    ascending=False
).reset_index(drop=True)

predictions_df["rank"] = (
    predictions_df["predicted_5d_return"]
    .rank(
        ascending=False,
        method="first")
    .astype(int))

predictions_df = predictions_df[[
        "rank",
        "date",
        "ticker",
        "model",
        "predicted_5d_return",
        "predicted_5d_return_pct",
        "predicted_direction",
        "training_rows"]]

output_path = (
    PROJECT_ROOT
    / "results"
    / "model_comparison"
    / "portfolio_svr_latest_predictions.csv")

output_path.parent.mkdir(
    parents=True,
    exist_ok=True)

predictions_df.to_csv(
    output_path,
    index=False)

print("\n" + "=" * 80)
print("LATEST SVR PREDICTIONS")
print("=" * 80)

display(
    predictions_df.round(6))

print("\nSaved to:")
print(output_path)

print(
    "\nStatus: Portfolio-wide latest SVR prediction engine completed.")
```

```python
# PORTFOLIO — RISK SNAPSHOT
# Volatility, returns, drawdown and beta vs S&P 500

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

INPUT_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "features.parquet")

df = pd.read_parquet(INPUT_PATH)

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

# Extract adjusted close prices

price_data = pd.DataFrame(index=df.index)

for ticker in UNIVERSE:
    price_data[ticker] = df[("Adj Close", ticker)]

benchmark = df[("Adj Close", "^GSPC")].rename("SP500")

returns = price_data.pct_change()
benchmark_returns = benchmark.pct_change()

# Risk metrics

rows = []

for ticker in UNIVERSE:

    r = returns[ticker].dropna()

    # Align with benchmark for beta
    aligned = pd.concat(
        [r, benchmark_returns],
        axis=1
    ).dropna()

    stock_r = aligned[ticker]
    market_r = aligned["SP500"]

    # Annualized volatility
    annual_vol = stock_r.std() * np.sqrt(252)

    # Returns
    return_20d = (
        price_data[ticker].iloc[-1]
        / price_data[ticker].iloc[-21]
        - 1)

    return_60d = (
        price_data[ticker].iloc[-1]
        / price_data[ticker].iloc[-61]
        - 1)

    return_252d = (
        price_data[ticker].iloc[-1]
        / price_data[ticker].iloc[-253]
        - 1)

    # Current drawdown
    wealth = (1 + r).cumprod()
    running_max = wealth.cummax()
    drawdown = wealth / running_max - 1

    current_drawdown = drawdown.iloc[-1]
    max_drawdown = drawdown.min()

    # Beta vs S&P 500
    market_variance = market_r.var()

    if market_variance > 0:
        beta = (
            stock_r.cov(market_r)
            / market_variance)
    else:
        beta = np.nan

    rows.append({
        "ticker": ticker,
        "latest_price": price_data[ticker].iloc[-1],
        "annualized_volatility": annual_vol,
        "return_20d": return_20d,
        "return_60d": return_60d,
        "return_252d": return_252d,
        "current_drawdown": current_drawdown,
        "max_drawdown": max_drawdown,
        "beta_vs_sp500": beta})

risk_df = pd.DataFrame(rows)

risk_df = risk_df.sort_values(
    "annualized_volatility"
).reset_index(drop=True)

output_path = (
    PROJECT_ROOT
    / "results"
    / "model_comparison"
    / "portfolio_risk_snapshot.csv")

output_path.parent.mkdir(
    parents=True,
    exist_ok=True)

risk_df.to_csv(
    output_path,
    index=False)

print("=" * 80)
print("PORTFOLIO — RISK SNAPSHOT")
print("=" * 80)

print(f"\nLatest date: {df.index.max()}")

display(
    risk_df.round(6))

print("\nSaved to:")
print(output_path)

print(
    "\nStatus: Portfolio-wide risk snapshot completed.")
```

```python
# PORTFOLIO — RESEARCH SIGNAL ENGINE
# Combine SVR prediction + model evidence + risk metrics

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

RESULTS_DIR = PROJECT_ROOT / "results"

PREDICTION_PATH = (
    RESULTS_DIR
    / "model_comparison"
    / "portfolio_svr_latest_predictions.csv")

RISK_PATH = (
    RESULTS_DIR
    / "model_comparison"
    / "portfolio_risk_snapshot.csv")

SVR_RESULTS_PATH = (
    RESULTS_DIR
    / "classical_ml"
    / "portfolio_svr_regression.csv")

# Load datasets

pred_df = pd.read_csv(
    PREDICTION_PATH)

risk_df = pd.read_csv(
    RISK_PATH)

svr_df = pd.read_csv(
    SVR_RESULTS_PATH)

# Keep SVR test evidence

svr_evidence = svr_df[
    [
        "ticker",
        "test_RMSE",
        "test_MAE",
        "test_R2",
        "test_directional_accuracy"]
].copy()

# Merge

signal_df = pred_df.merge(
    risk_df,
    on="ticker",
    how="left"
).merge(
    svr_evidence,
    on="ticker",
    how="left")

# Prediction strength

signal_df["prediction_strength"] = (
    signal_df["predicted_5d_return"]
    / signal_df["annualized_volatility"])

# Risk state

def risk_state(row):

    vol = row["annualized_volatility"]
    drawdown = row["current_drawdown"]
    beta = row["beta_vs_sp500"]

    if (
        vol >= 0.40
        or drawdown <= -0.30
        or beta >= 1.50):
        return "HIGH"

    elif (
        vol >= 0.25
        or drawdown <= -0.15
        or beta >= 1.20):
        return "MODERATE"

    else:
        return "LOW"


signal_df["risk_state"] = signal_df.apply(
    risk_state,
    axis=1)

# Momentum state

def momentum_state(row):

    r20 = row["return_20d"]
    r60 = row["return_60d"]
    r252 = row["return_252d"]

    positive_count = sum([
        r20 > 0,
        r60 > 0,
        r252 > 0])

    negative_count = sum([
        r20 < 0,
        r60 < 0,
        r252 < 0])

    if positive_count >= 2:
        return "POSITIVE"

    elif negative_count >= 2:
        return "NEGATIVE"

    else:
        return "MIXED"


signal_df["momentum_state"] = signal_df.apply(
    momentum_state,
    axis=1)

# Research signal classification
def research_signal(row):

    predicted_return = row[
        "predicted_5d_return"]

    risk = row["risk_state"]

    momentum = row["momentum_state"]

    direction = row[
        "predicted_direction"]

    # Strong positive prediction with acceptable risk
    if (
        predicted_return >= 0.01
        and direction == "POSITIVE"
        and risk != "HIGH"
        and momentum == "POSITIVE"):
        return "POSITIVE_RESEARCH_SIGNAL"

    # Positive prediction but risk/momentum conflict
    elif (
        predicted_return > 0
        and direction == "POSITIVE"):
        return "CAUTIOUS_POSITIVE"

    elif predicted_return <= 0:
        return "NON_POSITIVE"

    else:
        return "NEUTRAL"

    
signal_df["research_signal"] = signal_df.apply(
    research_signal,
    axis=1)

# Reason codes

def reason_codes(row):

    reasons = []

    if row["predicted_5d_return"] > 0:
        reasons.append("SVR_POSITIVE")
    else:
        reasons.append("SVR_NON_POSITIVE")

    if row["risk_state"] == "HIGH":
        reasons.append("HIGH_RISK")
    elif row["risk_state"] == "MODERATE":
        reasons.append("MODERATE_RISK")
    else:
        reasons.append("LOW_RISK")

    if row["momentum_state"] == "POSITIVE":
        reasons.append("POSITIVE_MOMENTUM")
    elif row["momentum_state"] == "NEGATIVE":
        reasons.append("NEGATIVE_MOMENTUM")
    else:
        reasons.append("MIXED_MOMENTUM")

    if row["test_directional_accuracy"] >= 55:
        reasons.append("TEST_DIRECTIONAL_EVIDENCE_55PCT_PLUS")

    if row["test_R2"] >= 0:
        reasons.append("NON_NEGATIVE_TEST_R2")

    return "|".join(reasons)


signal_df["reason_codes"] = signal_df.apply(
    reason_codes,
    axis=1)

# Rank research signals

signal_df["research_rank"] = (
    signal_df["predicted_5d_return"]
    .rank(
        ascending=False,
        method="first"
    )
    .astype(int))

signal_df = signal_df.sort_values(
    "research_rank"
).reset_index(drop=True)

# Select final columns

signal_df = signal_df[
    [
        "research_rank",
        "date",
        "ticker",
        "model",
        "predicted_5d_return",
        "predicted_5d_return_pct",
        "predicted_direction",
        "prediction_strength",
        "test_RMSE",
        "test_MAE",
        "test_R2",
        "test_directional_accuracy",
        "annualized_volatility",
        "return_20d",
        "return_60d",
        "return_252d",
        "current_drawdown",
        "max_drawdown",
        "beta_vs_sp500",
        "risk_state",
        "momentum_state",
        "research_signal",
        "reason_codes"]]

output_path = (
    RESULTS_DIR
    / "model_comparison"
    / "portfolio_research_signals.csv")

signal_df.to_csv(
    output_path,
    index=False)

print("=" * 80)
print("PORTFOLIO — RESEARCH SIGNAL ENGINE")
print("=" * 80)

print(
    f"\nLatest date: "
    f"{signal_df['date'].iloc[0]}")

display(
    signal_df.round(6))

print("\nSaved to:")
print(output_path)

print(
    "\nStatus: Portfolio research signal engine completed.")
```

```python
# PORTFOLIO — SIGNAL-AWARE MPT OPTIMIZATION
# Combine SVR expected returns with historical covariance
# Long-only portfolio with maximum weight constraint

import pandas as pd
import numpy as np
from pathlib import Path
from scipy.optimize import minimize

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

FEATURES_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "features.parquet")

SIGNAL_PATH = (
    PROJECT_ROOT
    / "results"
    / "model_comparison"
    / "portfolio_research_signals.csv")

OUTPUT_DIR = (
    PROJECT_ROOT
    / "results"
    / "portfolio")

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True)

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

# Load historical prices

df = pd.read_parquet(
    FEATURES_PATH)

price_data = pd.DataFrame(
    index=df.index)

for ticker in UNIVERSE:
    price_data[ticker] = df[
        ("Adj Close", ticker)]

price_data = price_data.dropna()

daily_returns = price_data.pct_change().dropna()

# Annualized covariance matrix

cov_matrix = (
    daily_returns.cov()
    * 252)

# Load SVR research signals

signals = pd.read_csv(
    SIGNAL_PATH)

signals = signals[
    signals["ticker"].isin(UNIVERSE)
].copy()

signals = signals.set_index(
    "ticker"
).loc[UNIVERSE]

predicted_returns = signals[
    "predicted_5d_return"
].values

# Convert 5-day prediction to approximate annualized
# expected return for the MPT objective.

# This is a simple scaling assumption:
# approximately 252 / 5 five-day periods per year.

expected_returns = (
    predicted_returns
    * (252 / 5))

# Portfolio mathematics

def portfolio_return(weights):
    return np.dot(
        weights,
        expected_returns)

def portfolio_volatility(weights):

    return np.sqrt(
        np.dot(
            weights.T,
            np.dot(
                cov_matrix.values,
                weights)))

def portfolio_sharpe(weights):

    vol = portfolio_volatility(
        weights)

    if vol == 0:
        return -np.inf

    return (
        portfolio_return(weights)
        / vol)

# Optimization

n_assets = len(UNIVERSE)

initial_weights = np.ones(
    n_assets
) / n_assets

constraints = [
    {
        "type": "eq",
        "fun": lambda w: np.sum(w) - 1}]

# Maximum 30% per individual stock
bounds = [
    (0.0, 0.30)
    for _ in range(n_assets)]

optimization = minimize(
    lambda w: -portfolio_sharpe(w),
    initial_weights,
    method="SLSQP",
    bounds=bounds,
    constraints=constraints,
    options={
        "maxiter": 1000,
        "ftol": 1e-10})

if not optimization.success:
    raise RuntimeError(
        "MPT optimization failed: "
        + optimization.message)

optimized_weights = optimization.x

# Equal-weight portfolio

equal_weights = (
    np.ones(n_assets)
    / n_assets)

# Build weight table

weights_df = pd.DataFrame({
    "ticker": UNIVERSE,
    "SVR_predicted_5d_return": predicted_returns,
    "annualized_expected_return": expected_returns,
    "optimized_weight": optimized_weights,
    "equal_weight": equal_weights})

weights_df["optimized_weight_pct"] = (
    weights_df["optimized_weight"] * 100)

weights_df["equal_weight_pct"] = (
    weights_df["equal_weight"] * 100)

weights_df = weights_df.sort_values(
    "optimized_weight",
    ascending=False
).reset_index(drop=True)

# Portfolio statistics

optimized_return = portfolio_return(
    optimized_weights)

optimized_vol = portfolio_volatility(
    optimized_weights)

optimized_sharpe = portfolio_sharpe(
    optimized_weights)

equal_return = portfolio_return(
    equal_weights)

equal_vol = portfolio_volatility(
    equal_weights)

equal_sharpe = portfolio_sharpe(
    equal_weights)

comparison_df = pd.DataFrame([
    {
        "portfolio": "Signal-Aware MPT",
        "expected_annual_return": optimized_return,
        "annualized_volatility": optimized_vol,
        "expected_sharpe": optimized_sharpe
    },
    {
        "portfolio": "Equal Weight",
        "expected_annual_return": equal_return,
        "annualized_volatility": equal_vol,
        "expected_sharpe": equal_sharpe
    }])

weights_path = (
    OUTPUT_DIR
    / "signal_aware_mpt_weights.csv")

comparison_path = (
    OUTPUT_DIR
    / "signal_aware_mpt_comparison.csv")

weights_df.to_csv(
    weights_path,
    index=False)

comparison_df.to_csv(
    comparison_path,
    index=False)

print("=" * 80)
print("SIGNAL-AWARE MPT OPTIMIZATION")
print("=" * 80)

print(
    f"\nSignal date: "
    f"{signals['date'].iloc[0]}")

print(
    f"Assets: {n_assets}")

print(
    f"Maximum individual weight: 30%")

print("\nOptimization status:")
print(optimization.message)

print("\nPortfolio weights:")
display(
    weights_df.round(6))

print("\nPortfolio comparison:")
display(
    comparison_df.round(6))

print("\nSaved weights to:")
print(weights_path)

print("\nSaved comparison to:")
print(comparison_path)

print(
    "\nStatus: Signal-aware MPT optimization completed.")
```

```python
# SIGNAL-AWARE MPT — ROBUST VERSION
# Use 5-day model signals with conservative portfolio tilts

import pandas as pd
import numpy as np
from pathlib import Path
from scipy.optimize import minimize

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

FEATURES_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "features.parquet")

SIGNAL_PATH = (
    PROJECT_ROOT
    / "results"
    / "model_comparison"
    / "portfolio_research_signals.csv")

OUTPUT_DIR = (
    PROJECT_ROOT
    / "results"
    / "portfolio")

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True)

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

df = pd.read_parquet(
    FEATURES_PATH)

prices = pd.DataFrame(index=df.index)

for ticker in UNIVERSE:
    prices[ticker] = df[
        ("Adj Close", ticker)]

prices = prices.dropna()

daily_returns = (
    prices.pct_change()
    .dropna())

# Annualized covariance
cov_matrix = (
    daily_returns.cov() * 252)

# Load latest SVR signals

signals = pd.read_csv(
    SIGNAL_PATH)

signals = signals[
    signals["ticker"].isin(UNIVERSE)
].copy()

signals = signals.set_index(
    "ticker"
).loc[UNIVERSE]

predicted_5d = (
    signals["predicted_5d_return"]
    .astype(float)
    .values)

# Conservative signal transformation
#
# Instead of annualizing the raw 5-day prediction,
# convert the prediction into a bounded relative score.

signal_mean = predicted_5d.mean()
signal_std = predicted_5d.std()

if signal_std == 0:
    signal_score = np.zeros(
        len(predicted_5d))
else:
    signal_score = (
        (predicted_5d - signal_mean)
        / signal_std)

# Cap extreme signal influence
signal_score = np.clip(
    signal_score,
    -2.0,
    2.0)

# Portfolio objective

# Base = equal-weight portfolio.
# Signals create controlled tilts around equal weight.
# Risk penalty discourages excessive volatility.

n_assets = len(UNIVERSE)

equal_weights = (
    np.ones(n_assets)
    / n_assets)

# Conservative signal coefficient
SIGNAL_STRENGTH = 0.025

# Risk-aversion coefficient
RISK_AVERSION = 1.0

def portfolio_volatility(weights):

    return np.sqrt(
        weights.T
        @ cov_matrix.values
        @ weights)

def objective(weights):

    signal_component = np.dot(
        weights,
        signal_score)

    risk_component = (
        portfolio_volatility(weights)
        ** 2)

    return -(
        SIGNAL_STRENGTH
        * signal_component
        - RISK_AVERSION
        * risk_component)

# Constraints

constraints = [
    {
        "type": "eq",
        "fun": lambda w: np.sum(w) - 1}]

# Maximum 20% per stock
bounds = [
    (0.0, 0.20)
    for _ in range(n_assets)]

optimization = minimize(
    objective,
    equal_weights,
    method="SLSQP",
    bounds=bounds,
    constraints=constraints,
    options={
        "maxiter": 2000,
        "ftol": 1e-12})

if not optimization.success:
    raise RuntimeError(
        "Optimization failed: "
        + optimization.message)

optimized_weights = optimization.x

# Portfolio statistics

optimized_vol = portfolio_volatility(
    optimized_weights)

equal_vol = portfolio_volatility(
    equal_weights)

# Signal exposure
optimized_signal_exposure = np.dot(
    optimized_weights,
    predicted_5d)

equal_signal_exposure = np.dot(
    equal_weights,
    predicted_5d)

# Weight table

weights_df = pd.DataFrame({
    "ticker": UNIVERSE,
    "predicted_5d_return": predicted_5d,
    "signal_score": signal_score,
    "optimized_weight": optimized_weights,
    "equal_weight": equal_weights})

weights_df["optimized_weight_pct"] = (
    weights_df["optimized_weight"] * 100)

weights_df["equal_weight_pct"] = (
    weights_df["equal_weight"] * 100)

weights_df = weights_df.sort_values(
    "optimized_weight",
    ascending=False
).reset_index(drop=True)

# Comparison

comparison_df = pd.DataFrame([
    {
        "portfolio": "Robust Signal-Aware MPT",
        "annualized_volatility": optimized_vol,
        "weighted_predicted_5d_return":
            optimized_signal_exposure
    },
    {
        "portfolio": "Equal Weight",
        "annualized_volatility": equal_vol,
        "weighted_predicted_5d_return":
            equal_signal_exposure}])


weights_path = (
    OUTPUT_DIR
    / "robust_signal_aware_mpt_weights.csv")

comparison_path = (
    OUTPUT_DIR
    / "robust_signal_aware_mpt_comparison.csv")

weights_df.to_csv(
    weights_path,
    index=False)

comparison_df.to_csv(
    comparison_path,
    index=False)

print("=" * 80)
print("ROBUST SIGNAL-AWARE MPT")
print("=" * 80)

print(
    f"\nSignal date: "
    f"{signals['date'].iloc[0]}")

print(
    "\nOptimization status:")

print(
    optimization.message)

print(
    f"\nSignal strength: {SIGNAL_STRENGTH}")

print(
    f"Risk aversion: {RISK_AVERSION}")

print(
    "\nPortfolio weights:")

display(
    weights_df.round(6))

print(
    "\nPortfolio comparison:")

display(
    comparison_df.round(6))

print(
    "\nSaved weights to:")

print(weights_path)

print(
    "\nSaved comparison to:")

print(comparison_path)

print(
    "\nStatus: Robust signal-aware MPT completed.")
```

```python
# OOS VALIDATION — ROBUST SIGNAL-AWARE PORTFOLIO
# Compare against Equal Weight and S&P 500

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

FEATURES_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "features.parquet")

WEIGHTS_PATH = (
    PROJECT_ROOT
    / "results"
    / "portfolio"
    / "robust_signal_aware_mpt_weights.csv")

OUTPUT_DIR = (
    PROJECT_ROOT
    / "results"
    / "portfolio")

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True)

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

df = pd.read_parquet(
    FEATURES_PATH)

prices = pd.DataFrame(index=df.index)

for ticker in UNIVERSE:
    prices[ticker] = df[
        ("Adj Close", ticker)]

sp500 = df[
    ("Adj Close", "^GSPC")
].rename("SP500")

prices = prices.dropna()
sp500 = sp500.loc[prices.index]

daily_returns = (
    prices.pct_change()
    .dropna())

sp500_returns = (
    sp500.pct_change()
    .loc[daily_returns.index])

# Use the previously validated chronological OOS window

OOS_START = pd.Timestamp("2025-01-10")
OOS_END = daily_returns.index.max()

oos_returns = daily_returns.loc[
    OOS_START:OOS_END]

oos_sp500 = sp500_returns.loc[
    OOS_START:OOS_END]

# Load current robust MPT weights

weights_df = pd.read_csv(
    WEIGHTS_PATH)

weights_df = weights_df.set_index(
    "ticker"
).loc[UNIVERSE]

weights = (
    weights_df[
        "optimized_weight"
    ]
    .astype(float)
    .values)

equal_weights = (
    np.ones(len(UNIVERSE))
    / len(UNIVERSE))

# Portfolio daily returns

signal_portfolio_returns = (
    oos_returns
    .dot(weights))

equal_portfolio_returns = (
    oos_returns
    .dot(equal_weights))

# Wealth curves

initial_capital = 100000

signal_wealth = (
    initial_capital
    * (1 + signal_portfolio_returns)
    .cumprod())

equal_wealth = (
    initial_capital
    * (1 + equal_portfolio_returns)
    .cumprod())

sp500_wealth = (
    initial_capital
    * (1 + oos_sp500)
    .cumprod())

# Metrics helper

def calculate_metrics(
    returns,
    wealth,
    benchmark_returns):

    returns = returns.dropna()

    total_return = (
        wealth.iloc[-1]
        / initial_capital
        - 1)

    years = (
        len(returns) / 252)

    annualized_return = (
        (1 + total_return)
        ** (1 / years)
        - 1)

    annualized_volatility = (
        returns.std()
        * np.sqrt(252))

    sharpe = (
        annualized_return
        / annualized_volatility
        if annualized_volatility > 0
        else np.nan)

    downside = returns[
        returns < 0]

    downside_volatility = (
        downside.std()
        * np.sqrt(252)
        if len(downside) > 1
        else np.nan)

    sortino = (
        annualized_return
        / downside_volatility
        if (
            downside_volatility
            and downside_volatility > 0)
        else np.nan)

    running_max = wealth.cummax()

    drawdown = (
        wealth
        / running_max
        - 1)

    max_drawdown = drawdown.min()

    aligned = pd.concat(
        [
            returns.rename("portfolio"),
            benchmark_returns.rename("benchmark")
        ],
        axis=1
    ).dropna()

    beta = (
        aligned["portfolio"].cov(
            aligned["benchmark"]
        )
        / aligned["benchmark"].var())

    return {
        "total_return": total_return,
        "annualized_return": annualized_return,
        "annualized_volatility":
            annualized_volatility,
        "sharpe": sharpe,
        "sortino": sortino,
        "max_drawdown": max_drawdown,
        "beta_vs_sp500": beta,
        "final_value": wealth.iloc[-1]}

# Calculate metrics

signal_metrics = calculate_metrics(
    signal_portfolio_returns,
    signal_wealth,
    oos_sp500)

equal_metrics = calculate_metrics(
    equal_portfolio_returns,
    equal_wealth,
    oos_sp500)

sp500_metrics = calculate_metrics(
    oos_sp500,
    sp500_wealth,
    oos_sp500)

# Comparison table

comparison_df = pd.DataFrame([
    {
        "portfolio":
            "Robust Signal-Aware MPT",
        **signal_metrics
    },
    {
        "portfolio":
            "Equal Weight",
        **equal_metrics
    },
    {
        "portfolio":
            "S&P 500",
        **sp500_metrics
    }])

# Save wealth series

wealth_df = pd.DataFrame({
    "date": oos_returns.index,
    "signal_aware_mpt":
        signal_wealth.values,
    "equal_weight":
        equal_wealth.values,
    "sp500":
        sp500_wealth.values})

metrics_path = (
    OUTPUT_DIR
    / "robust_signal_aware_mpt_oos_metrics.csv")

wealth_path = (
    OUTPUT_DIR
    / "robust_signal_aware_mpt_oos_wealth.csv")

comparison_df.to_csv(
    metrics_path,
    index=False)

wealth_df.to_csv(
    wealth_path,
    index=False)

# Display

print("=" * 80)
print("ROBUST SIGNAL-AWARE MPT — OOS VALIDATION")
print("=" * 80)

print(
    f"\nOOS period: "
    f"{oos_returns.index.min().date()} "
    f"→ "
    f"{oos_returns.index.max().date()}")

print(
    f"Initial capital: "
    f"₹{initial_capital:,.0f}")

print("\nPortfolio weights used:")
display(
    weights_df[
        [
            "predicted_5d_return",
            "optimized_weight"
        ]
    ].round(6))

print("\nOOS performance:")
display(
    comparison_df.round(6))

print("\nSaved metrics to:")
print(metrics_path)

print("\nSaved wealth series to:")
print(wealth_path)

print(
    "\nStatus: Robust signal-aware MPT OOS validation completed.")
```

## NORTHGATE AI — WALK-FORWARD REBALANCING ENGINE

```python
# Strict chronological protocol:

# At each rebalance date:
# 1. Use ONLY data available before rebalance date
# 2. Train one SVR per stock
# 3. Predict next 5-day return
# 4. Apply risk-aware signal transformation
# 5. Optimize portfolio weights
# 6. Hold until next rebalance
# 7. Apply transaction costs
 
#####  No future data is used to generate historical signals.
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.optimize import minimize

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR


PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

FEATURES_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "features.parquet")

SCHEDULE_PATH = (
    PROJECT_ROOT
    / "results"
    / "portfolio"
    / "walk_forward_rebalance_schedule_corrected.csv")

OUTPUT_DIR = (
    PROJECT_ROOT
    / "results"
    / "portfolio")

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True)

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

# Configuration

INITIAL_CAPITAL = 100000.0

MAX_WEIGHT = 0.20

SIGNAL_STRENGTH = 0.025

RISK_AVERSION = 1.0

TRANSACTION_COST = 0.001   # 0.10%

LOOKBACK_DAYS = 1260       # approximately 5 years

MIN_TRAIN_ROWS = 500


# LOAD DATA

print("=" * 80)
print("NORTHGATE AI — WALK-FORWARD REBALANCING ENGINE")
print("=" * 80)

df = pd.read_parquet(
    FEATURES_PATH)

df = df.sort_index()

prices = pd.DataFrame(
    index=df.index)

for ticker in UNIVERSE:

    prices[ticker] = df[
        ("Adj Close", ticker)]

prices = prices.dropna()

daily_returns = (
    prices
    .pct_change()
    .dropna())

schedule_df = pd.read_csv(
    SCHEDULE_PATH,
    parse_dates=[
        "rebalance_date",
        "holding_end_date"])

print(
    f"\nRebalance periods: "
    f"{len(schedule_df)}")

print(
    f"Initial capital: "
    f"₹{INITIAL_CAPITAL:,.0f}")

print(
    f"Transaction cost: "
    f"{TRANSACTION_COST * 100:.2f}%")

# FEATURE MATRIX

feature_columns = [
    col
    for col in df.columns
    if not (
        "future_return_5d_" in str(col)
        or "target_5d_" in str(col))]

X_all = df[
    feature_columns
].select_dtypes(
    include=[np.number]
).copy()

# Align with prices
X_all = X_all.loc[
    prices.index]

# HELPER — MPT OPTIMIZATION

def optimize_weights(
    predicted_returns,
    covariance
):

    n_assets = len(UNIVERSE)

    equal_weights = (
        np.ones(n_assets)
        / n_assets)

    mean_signal = np.mean(
        predicted_returns)

    std_signal = np.std(
        predicted_returns)

    if std_signal > 0:

        signal_score = (
            predicted_returns
            - mean_signal
        ) / std_signal

    else:

        signal_score = np.zeros(
            n_assets)

    signal_score = np.clip(
        signal_score,
        -2,
        2)

    def portfolio_volatility(w):

        variance = (
            w.T
            @ covariance
            @ w)

        return np.sqrt(
            max(variance, 0))

    def objective(w):

        signal_component = np.dot(
            w,
            signal_score)

        risk_component = (
            portfolio_volatility(w)
            ** 2)

        return -(
            SIGNAL_STRENGTH
            * signal_component
            -
            RISK_AVERSION
            * risk_component)

    constraints = [
        {
            "type": "eq",
            "fun": lambda w:
                np.sum(w) - 1}]

    bounds = [
        (0.0, MAX_WEIGHT)
        for _ in range(n_assets)]

    result = minimize(
        objective,
        equal_weights,
        method="SLSQP",
        bounds=bounds,
        constraints=constraints,
        options={
            "maxiter": 1000,
            "ftol": 1e-10})

    if not result.success:

        return equal_weights

    return result.x

# WALK-FORWARD LOOP

rebalance_records = []

daily_portfolio_records = []

previous_weights = np.zeros(
    len(UNIVERSE))

current_capital = INITIAL_CAPITAL


for idx, row in schedule_df.iterrows():

    rebalance_date = pd.Timestamp(
        row["rebalance_date"])

    holding_end = pd.Timestamp(
        row["holding_end_date"])

    print(
        f"\n[{idx + 1}/{len(schedule_df)}]"
        f" Rebalance: "
        f"{rebalance_date.date()}")

    # Historical training data
    # ONLY dates strictly before rebalance date

    historical_dates = X_all.index[
        X_all.index < rebalance_date]

    if len(historical_dates) < MIN_TRAIN_ROWS:

        print(
            "  Skipped — insufficient "
            "historical training data.")

        continue

    historical_dates = historical_dates[
        -LOOKBACK_DAYS:]

    X_train_full = X_all.loc[
        historical_dates]

    # Covariance from historical data only

    historical_returns = daily_returns.loc[
        daily_returns.index < rebalance_date
    ].tail(
        LOOKBACK_DAYS)

    covariance = (
        historical_returns
        .cov()
        .values
        * 252)

    # Latest feature row
    # Available at rebalance date

    feature_candidates = X_all.index[
        X_all.index <= rebalance_date]

    if len(feature_candidates) == 0:

        continue

    prediction_date = (
        feature_candidates[-1])

    X_latest = X_all.loc[
        [prediction_date]]

    predicted_returns = []

    # Train one SVR per stock

    for ticker in UNIVERSE:

        target_column = (
            f"future_return_5d_{ticker}")

        # Construct target from price history
        #
        # Target at date t:
        # future 5-day return after t.
        target_series = (
            prices[ticker]
            .shift(-5)
            / prices[ticker]
            - 1)

        y_train = target_series.loc[
            X_train_full.index]

        valid_mask = y_train.notna()

        Xtr = X_train_full.loc[
            valid_mask]

        ytr = y_train.loc[
            valid_mask]

        if len(ytr) < MIN_TRAIN_ROWS:

            predicted_returns.append(
                0.0)

            continue

        model = Pipeline([
            (
                "imputer",
                SimpleImputer(
                    strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            ),
            (
                "model",
                SVR(
                    C=0.1,
                    gamma="scale",
                    epsilon=0.01,
                    kernel="rbf"))])

        model.fit(
            Xtr,
            ytr.astype(float))

        prediction = float(
            model.predict(
                X_latest
            )[0])

        predicted_returns.append(
            prediction)

    predicted_returns = np.asarray(
        predicted_returns)

    # Optimize weights

    weights = optimize_weights(
        predicted_returns,
        covariance)

    # Transaction cost
    # Turnover = sum(abs(new - old))

    turnover = np.sum(
        np.abs(
            weights
            - previous_weights))

    transaction_cost = (
        turnover
        * TRANSACTION_COST)

    # Holding-period returns

    period_returns = daily_returns.loc[
        (daily_returns.index >= rebalance_date)
        &
        (daily_returns.index <= holding_end)]

    if len(period_returns) == 0:

        continue

    portfolio_daily_returns = (
        period_returns
        .dot(weights))

    # Apply transaction cost on first day
    # of each holding period.
    portfolio_daily_returns = (
        portfolio_daily_returns
        .copy())

    portfolio_daily_returns.iloc[0] -= (
        transaction_cost)

    # Store daily results
    for date, daily_return in (
        portfolio_daily_returns.items()
    ):

        daily_portfolio_records.append({
            "date": date,
            "rebalance_number":
                idx + 1,
            "rebalance_date":
                rebalance_date,
            "portfolio_return":
                daily_return})

    # Store rebalance-level results

    rebalance_records.append({
        "rebalance_number":
            idx + 1,
        "rebalance_date":
            rebalance_date,
        "prediction_date":
            prediction_date,
        "holding_end_date":
            holding_end,
        "turnover":
            turnover,
        "transaction_cost":
            transaction_cost,
        "predicted_mean_5d_return":
            np.mean(predicted_returns),
        "predicted_max_5d_return":
            np.max(predicted_returns),
        "predicted_min_5d_return":
            np.min(predicted_returns)})

    # Save weights
    for ticker, weight, prediction in zip(
        UNIVERSE,
        weights,
        predicted_returns):

        rebalance_records[-1][
            f"weight_{ticker}"
        ] = weight

        rebalance_records[-1][
            f"prediction_{ticker}"
        ] = prediction

    previous_weights = weights.copy()

    print(
        f"  Prediction mean: "
        f"{np.mean(predicted_returns) * 100:.3f}%")

    print(
        f"  Turnover: "
        f"{turnover * 100:.2f}%")

    print(
        f"  Transaction cost: "
        f"{transaction_cost * 100:.3f}%")

    print(
        "  Weights:")

    for ticker, weight in zip(
        UNIVERSE,
        weights):

        if weight > 0.001:

            print(
                f"    {ticker}: "
                f"{weight * 100:.2f}%")


rebalance_results_df = pd.DataFrame(
    rebalance_records)

daily_results_df = pd.DataFrame(
    daily_portfolio_records)

rebalance_path = (
    OUTPUT_DIR
    / "walk_forward_rebalance_results.csv")

daily_path = (
    OUTPUT_DIR
    / "walk_forward_daily_returns.csv")

rebalance_results_df.to_csv(
    rebalance_path,
    index=False)

daily_results_df.to_csv(
    daily_path,
    index=False)

# FINAL SUMMARY

print("\n" + "=" * 80)
print("WALK-FORWARD REBALANCING ENGINE COMPLETE")
print("=" * 80)

print(
    f"\nCompleted rebalance periods: "
    f"{len(rebalance_results_df)}")

print(
    f"Daily observations: "
    f"{len(daily_results_df)}")

print("\nFirst rebalance results:")

display(
    rebalance_results_df.head(3).round(6))

print("\nLast rebalance results:")

display(
    rebalance_results_df.tail(3).round(6))

print("\nSaved rebalance results to:")
print(rebalance_path)

print("\nSaved daily returns to:")
print(daily_path)

print(
    "\nStatus: "
    "Walk-forward rebalancing engine completed.")
```

```python

# WALK-FORWARD PORTFOLIO — PERFORMANCE & RISK AUDIT

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

FEATURES_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "features.parquet")

DAILY_PATH = (
    PROJECT_ROOT
    / "results"
    / "portfolio"
    / "walk_forward_daily_returns.csv")

OUTPUT_DIR = (
    PROJECT_ROOT
    / "results"
    / "portfolio")

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True)

UNIVERSE = [
    "AAPL",
    "MSFT",
    "JPM",
    "XOM",
    "JNJ",
    "PG",
    "NVDA",
    "KO",
    "CAT",
    "HD"]

INITIAL_CAPITAL = 100000.0

# LOAD WALK-FORWARD RETURNS

wf = pd.read_csv(
    DAILY_PATH,
    parse_dates=["date"])

wf = wf.sort_values(
    "date"
).reset_index(drop=True)

wf_returns = (
    wf
    .set_index("date")[
        "portfolio_return"
    ]
    .astype(float))

df = pd.read_parquet(
    FEATURES_PATH)

prices = pd.DataFrame(
    index=df.index)

for ticker in UNIVERSE:

    prices[ticker] = df[
        ("Adj Close", ticker)]

sp500 = df[
    ("Adj Close", "^GSPC")
].rename("SP500")

prices = prices.dropna()

sp500 = sp500.loc[
    prices.index]

market_returns = (
    sp500
    .pct_change()
    .dropna())

# Align market to walk-forward period
market_returns = market_returns.loc[
    wf_returns.index]

# EQUAL-WEIGHT BENCHMARK

stock_returns = (
    prices
    .pct_change()
    .loc[wf_returns.index])

equal_weights = (
    np.ones(len(UNIVERSE))
    / len(UNIVERSE))

equal_returns = (
    stock_returns
    .dot(equal_weights))

# WEALTH CURVES

wf_wealth = (
    INITIAL_CAPITAL
    * (1 + wf_returns)
    .cumprod())

equal_wealth = (
    INITIAL_CAPITAL
    * (1 + equal_returns)
    .cumprod())

sp500_wealth = (
    INITIAL_CAPITAL
    * (1 + market_returns)
    .cumprod())

# METRICS FUNCTION

def portfolio_metrics(
    returns,
    wealth,
    benchmark):

    returns = returns.dropna()

    total_return = (
        wealth.iloc[-1]
        / INITIAL_CAPITAL
        - 1)

    years = (
        len(returns) / 252)

    annualized_return = (
        (1 + total_return)
        ** (1 / years)
        - 1)

    volatility = (
        returns.std()
        * np.sqrt(252))

    sharpe = (
        annualized_return
        / volatility
        if volatility > 0
        else np.nan)

    negative_returns = returns[
        returns < 0]

    downside_vol = (
        negative_returns.std()
        * np.sqrt(252)
        if len(negative_returns) > 1
        else np.nan)

    sortino = (
        annualized_return
        / downside_vol
        if (
            downside_vol
            and downside_vol > 0)
        else np.nan)

    running_max = (
        wealth
        .cummax())

    drawdown = (
        wealth
        / running_max
        - 1)

    max_drawdown = (
        drawdown.min())

    aligned = pd.concat(
        [
            returns.rename("portfolio"),
            benchmark.rename("benchmark")
        ],
        axis=1,
        join="inner"
    ).dropna()

    beta = (
        aligned["portfolio"].cov(
            aligned["benchmark"]
        )
        / aligned["benchmark"].var())

    return {
        "total_return": total_return,
        "annualized_return": annualized_return,
        "annualized_volatility": volatility,
        "sharpe": sharpe,
        "sortino": sortino,
        "max_drawdown": max_drawdown,
        "beta_vs_sp500": beta,
        "final_value": wealth.iloc[-1]}

# CALCULATE ALL METRICS

wf_metrics = portfolio_metrics(
    wf_returns,
    wf_wealth,
    market_returns)

equal_metrics = portfolio_metrics(
    equal_returns,
    equal_wealth,
    market_returns)

sp500_metrics = portfolio_metrics(
    market_returns,
    sp500_wealth,
    market_returns)

# COMPARISON TABLE

comparison_df = pd.DataFrame([
    {
        "portfolio":
            "Walk-Forward Signal-Aware MPT",
        **wf_metrics
    },
    {
        "portfolio":
            "Equal Weight",
        **equal_metrics
    },
    {
        "portfolio":
            "S&P 500",
        **sp500_metrics}])

# WEALTH SERIES

wealth_df = pd.DataFrame({
    "date": wf_returns.index,
    "walk_forward_mpt":
        wf_wealth.values,
    "equal_weight":
        equal_wealth.values,
    "sp500":
        sp500_wealth.values})

# SAVE RESULTS

metrics_path = (
    OUTPUT_DIR
    / "walk_forward_performance_metrics.csv")

wealth_path = (
    OUTPUT_DIR
    / "walk_forward_wealth_comparison.csv")

chart_path = (
    OUTPUT_DIR
    / "walk_forward_wealth_comparison.png")

comparison_df.to_csv(
    metrics_path,
    index=False)

wealth_df.to_csv(
    wealth_path,
    index=False)
```

```python
# WALK-FORWARD — EXECUTION TIMING AUDIT 

import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path(r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")
RESULTS_DIR = PROJECT_ROOT / "results" / "portfolio"

WF_RESULTS = RESULTS_DIR / "walk_forward_rebalance_results.csv"
WF_DAILY = RESULTS_DIR / "walk_forward_daily_returns.csv"

reb = pd.read_csv(WF_RESULTS, parse_dates=["rebalance_date"])
daily = pd.read_csv(WF_DAILY, parse_dates=["date"])

reb = reb.sort_values("rebalance_date").reset_index(drop=True)
daily = daily.sort_values("date").reset_index(drop=True)

print("=" * 90)
print("NORTHGATE AI — FIXED WALK-FORWARD EXECUTION TIMING AUDIT")
print("=" * 90)

# 1. Inspect actual columns

print("\n[1] DAILY RETURN FILE COLUMNS")
print(list(daily.columns))

print("\n[2] REBALANCE FILE COLUMNS")
print(list(reb.columns))

# 2. Coverage

print("\n[3] COVERAGE")
print(f"Rebalance periods : {len(reb)}")
print(f"Daily observations: {len(daily)}")
print(f"Start date        : {daily['date'].min().date()}")
print(f"End date          : {daily['date'].max().date()}")

# 3. Rebalance dates vs daily dates

reb_dates = set(reb["rebalance_date"].dt.normalize())
daily_dates = set(daily["date"].dt.normalize())

overlap = sorted(reb_dates.intersection(daily_dates))

print("\n[4] REBALANCE-DATE OVERLAP")
print(f"Overlapping rebalance dates: {len(overlap)} / {len(reb_dates)}")

print("\nFirst 10 rebalance dates:")
print([d.date() for d in overlap[:10]])

# 4. For every rebalance date, find the first daily observation
#    on or after that rebalance date

timing_rows = []

for _, row in reb.iterrows():

    rdate = pd.Timestamp(row["rebalance_date"]).normalize()

    future_dates = daily.loc[
        daily["date"].dt.normalize() >= rdate,
        "date"]

    if len(future_dates) == 0:
        first_daily = pd.NaT
    else:
        first_daily = future_dates.iloc[0]

    timing_rows.append({
        "rebalance_date": rdate,
        "first_daily_return_date": first_daily,
        "same_day_execution": (
            pd.notna(first_daily)
            and first_daily.normalize() == rdate),
        "days_between": (
            (first_daily.normalize() - rdate).days
            if pd.notna(first_daily) else None)})

timing_df = pd.DataFrame(timing_rows)

print("\n[5] EXECUTION TIMING")
print(
    f"Same-day starts: "
    f"{timing_df['same_day_execution'].sum()} / {len(timing_df)}")

print(
    f"Next-day-or-later starts: "
    f"{(~timing_df['same_day_execution']).sum()} / {len(timing_df)}")

display(timing_df)

# 5. Transaction cost / turnover audit

print("\n[6] TRANSACTION / TURNOVER COLUMNS")

transaction_cols = [
    c for c in daily.columns
    if any(word in c.lower() for word in ["cost", "turnover", "rebalance"])]

print(transaction_cols)

if transaction_cols:
    display(daily[["date"] + transaction_cols].head(15))

# 6. Return sanity

print("\n[7] RETURN SANITY CHECK")

portfolio_returns = daily["portfolio_return"].astype(float)

print(f"Mean daily return : {portfolio_returns.mean():.6%}")
print(f"Std daily return  : {portfolio_returns.std():.6%}")
print(f"Minimum return    : {portfolio_returns.min():.6%}")
print(f"Maximum return    : {portfolio_returns.max():.6%}")
print(f"Positive days     : {(portfolio_returns > 0).sum()}")
print(f"Negative days     : {(portfolio_returns < 0).sum()}")
print(f"Zero-return days  : {(portfolio_returns == 0).sum()}")

# 7. Save audit

AUDIT_PATH = RESULTS_DIR / "walk_forward_execution_timing_audit.csv"
timing_df.to_csv(AUDIT_PATH, index=False)

print("\n" + "=" * 90)
print("AUDIT COMPLETE")
print("=" * 90)
print(f"Saved timing audit to:\n{AUDIT_PATH}")
```

```python
# NORTHGATE AI — STRICT INFORMATION AVAILABILITY AUDIT

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

FEATURES_PATH = PROJECT_ROOT / "data" / "processed" / "features.parquet"
WF_RESULTS = PROJECT_ROOT / "results" / "portfolio" / "walk_forward_rebalance_results.csv"
WF_DAILY = PROJECT_ROOT / "results" / "portfolio" / "walk_forward_daily_returns.csv"

features = pd.read_parquet(FEATURES_PATH)
reb = pd.read_csv(
    WF_RESULTS,
    parse_dates=["rebalance_date", "prediction_date", "holding_end_date"])
daily = pd.read_csv(
    WF_DAILY,
    parse_dates=["date", "rebalance_date"])

reb = reb.sort_values("rebalance_number").reset_index(drop=True)
daily = daily.sort_values("date").reset_index(drop=True)

print("=" * 95)
print("NORTHGATE AI — STRICT INFORMATION AVAILABILITY AUDIT")
print("=" * 95)

# 1. Feature dataset date structure

feature_dates = pd.DatetimeIndex(features.index).sort_values()

print("\n[1] FEATURE DATA")
print(f"Feature start : {feature_dates.min().date()}")
print(f"Feature end   : {feature_dates.max().date()}")
print(f"Rows          : {len(feature_dates)}")

# 2. Check whether rebalance date itself exists in features

rebalance_dates = pd.DatetimeIndex(
    reb["rebalance_date"]
).normalize()

feature_date_set = set(feature_dates.normalize())

same_day_feature_available = [
    d in feature_date_set for d in rebalance_dates]

print("\n[2] SAME-DAY FEATURE AVAILABILITY")
print(
    f"Rebalance dates with same-day feature row: "
    f"{sum(same_day_feature_available)} / {len(rebalance_dates)}")

# 3. Show feature rows around first rebalance

first_rebalance = reb.loc[0, "rebalance_date"].normalize()

nearby_dates = feature_dates[
    (feature_dates >= first_rebalance - pd.Timedelta(days=3)) &
    (feature_dates <= first_rebalance + pd.Timedelta(days=3))]

print("\n[3] FEATURE DATES AROUND FIRST REBALANCE")
print([d.date() for d in nearby_dates])

# 4. Compare same-day vs next trading day

print("\n[4] EXECUTION DATE MAPPING")

mapping_rows = []

for _, row in reb.iterrows():

    rdate = pd.Timestamp(row["rebalance_date"]).normalize()

    available = feature_dates[feature_dates <= rdate]

    if len(available):
        latest_feature_date = available[-1]
    else:
        latest_feature_date = pd.NaT

    future_trading = daily.loc[
        daily["date"].dt.normalize() > rdate,
        "date"]

    if len(future_trading):
        next_trading_date = future_trading.iloc[0].normalize()
    else:
        next_trading_date = pd.NaT

    mapping_rows.append({
        "rebalance_number": row["rebalance_number"],
        "rebalance_date": rdate,
        "latest_feature_date_used_if_same_day": latest_feature_date,
        "feature_lag_days": (
            (rdate - latest_feature_date).days
            if pd.notna(latest_feature_date) else np.nan
        ),
        "next_trading_day": next_trading_date})

mapping = pd.DataFrame(mapping_rows)

display(mapping)

# 5. Inspect actual first-day portfolio return

print("\n[5] FIRST HOLDING-DAY RETURN CHECK")

first_days = (
    daily.sort_values(["rebalance_number", "date"])
    .groupby("rebalance_number", as_index=False)
    .first())

display(
    first_days[
        [
            "rebalance_number",
            "date",
            "rebalance_date",
            "portfolio_return"]
    ].head(10))

# 6. Determine whether first holding day == rebalance day

first_days["same_day_return"] = (
    first_days["date"].dt.normalize()
    == first_days["rebalance_date"].dt.normalize())

print(
    "\nSame-day portfolio return observations:",
    first_days["same_day_return"].sum(),
    "/",
    len(first_days))

# 7. Final diagnostic

print("\n" + "=" * 95)
print("STRICT TIMING DIAGNOSTIC")
print("=" * 95)

if (
    first_days["same_day_return"].all()
    and
    same_day_feature_available.count(True) == len(rebalance_dates)
):
    print(
        "⚠️ SAME-DAY SIGNAL + SAME-DAY RETURN CONFIRMED.")
    print(
        "The current backtest should NOT be treated as strictly "
        "next-period execution.")
    print(
        "Recommended correction: keep prediction at rebalance-date "
        "information, but start portfolio returns from the NEXT "
        "trading day.")
else:
    print(
        "Further inspection required before changing the execution model.")
```

```python
# NORTHGATE AI — STRICT NEXT-TRADING-DAY WALK-FORWARD BACKTEST
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

RESULTS_DIR = PROJECT_ROOT / "results" / "portfolio"
FEATURES_PATH = PROJECT_ROOT / "data" / "processed" / "features.parquet"
REB_PATH = RESULTS_DIR / "walk_forward_rebalance_results.csv"

OUTPUT_RESULTS = RESULTS_DIR / "strict_walk_forward_daily_returns.csv"
OUTPUT_SUMMARY = RESULTS_DIR / "strict_walk_forward_rebalance_audit.csv"

UNIVERSE = [
    "AAPL","MSFT","JPM","XOM","JNJ",
    "PG","NVDA","KO","CAT","HD"]

features = pd.read_parquet(FEATURES_PATH)
reb = pd.read_csv(
    REB_PATH,
    parse_dates=[
        "rebalance_date",
        "prediction_date",
        "holding_end_date"])

reb = reb.sort_values("rebalance_number").reset_index(drop=True)

# Build adjusted-close price panel

prices = pd.DataFrame(index=features.index)

for ticker in UNIVERSE:
    prices[ticker] = features[("Adj Close", ticker)]

prices = prices.dropna()

# Daily stock returns
stock_returns = prices.pct_change()

# Strict next-day execution

strict_rows = []
audit_rows = []

for i, row in reb.iterrows():

    rebalance_date = pd.Timestamp(row["rebalance_date"]).normalize()

    # First trading day STRICTLY AFTER rebalance date
    future_dates = stock_returns.index[
        stock_returns.index.normalize() > rebalance_date]

    if len(future_dates) == 0:
        continue

    first_return_date = future_dates[0]

    # Holding end
    holding_end = pd.Timestamp(row["holding_end_date"]).normalize()

    holding_dates = stock_returns.index[
        (stock_returns.index.normalize() >= first_return_date) &
        (stock_returns.index.normalize() <= holding_end)]

    if len(holding_dates) == 0:
        continue

    # Extract frozen weights from rebalance result
    weights = np.array([
        float(row[f"weight_{ticker}"])
        for ticker in UNIVERSE])

    # Portfolio return before transaction cost
    period_returns = stock_returns.loc[holding_dates, UNIVERSE]

    portfolio_returns = period_returns.dot(weights)

    # Transaction cost is applied ONLY on first day of new allocation
    transaction_cost = float(row["transaction_cost"])

    for j, date in enumerate(holding_dates):

        gross_return = float(portfolio_returns.loc[date])

        if j == 0:
            net_return = gross_return - transaction_cost
            cost_applied = transaction_cost
        else:
            net_return = gross_return
            cost_applied = 0.0

        strict_rows.append({
            "date": date,
            "rebalance_number": int(row["rebalance_number"]),
            "prediction_date": pd.Timestamp(
                row["prediction_date"]
            ).normalize(),
            "rebalance_date": rebalance_date,
            "first_return_date": first_return_date,
            "holding_end_date": holding_end,
            "gross_portfolio_return": gross_return,
            "transaction_cost": cost_applied,
            "portfolio_return": net_return})

    audit_rows.append({
        "rebalance_number": int(row["rebalance_number"]),
        "prediction_date": pd.Timestamp(
            row["prediction_date"]
        ).normalize(),
        "rebalance_date": rebalance_date,
        "first_return_date": first_return_date,
        "holding_end_date": holding_end,
        "same_day_execution": (
            first_return_date.normalize() == rebalance_date),
        "next_day_execution": (
            first_return_date.normalize() > rebalance_date),
        "turnover": float(row["turnover"]),
        "transaction_cost": transaction_cost,
        "predicted_mean_5d_return": float(
            row["predicted_mean_5d_return"])})

# Save outputs

strict_daily = pd.DataFrame(strict_rows)
strict_audit = pd.DataFrame(audit_rows)

strict_daily = (
    strict_daily
    .sort_values("date")
    .drop_duplicates(subset=["date"], keep="first")
    .reset_index(drop=True))

strict_audit.to_csv(OUTPUT_SUMMARY, index=False)
strict_daily.to_csv(OUTPUT_RESULTS, index=False)

# Validation

print("=" * 95)
print("NORTHGATE AI — STRICT NEXT-TRADING-DAY WALK-FORWARD BACKTEST")
print("=" * 95)

print("\n[1] COVERAGE")
print(f"Rebalance periods       : {len(strict_audit)}")
print(f"Daily observations      : {len(strict_daily)}")
print(f"Start performance date  : {strict_daily['date'].min().date()}")
print(f"End performance date    : {strict_daily['date'].max().date()}")

print("\n[2] EXECUTION TIMING")

print(
    "Same-day execution periods:",
    int(strict_audit["same_day_execution"].sum()))

print(
    "Next-day execution periods:",
    int(strict_audit["next_day_execution"].sum()))

print("\nFirst 10 execution mappings:")

display(
    strict_audit[
        ["rebalance_number",
            "prediction_date",
            "rebalance_date",
            "first_return_date",
            "holding_end_date",
            "same_day_execution",
            "next_day_execution",
            "transaction_cost"]
    ].head(10))

print("\n[3] RETURN SANITY")

r = strict_daily["portfolio_return"]

print(f"Mean daily return : {r.mean():.6%}")
print(f"Std daily return  : {r.std():.6%}")
print(f"Minimum return    : {r.min():.6%}")
print(f"Maximum return    : {r.max():.6%}")
print(f"Positive days     : {(r > 0).sum()}")
print(f"Negative days     : {(r < 0).sum()}")

print("\n[4] FILES SAVED")
print(OUTPUT_RESULTS)
print(OUTPUT_SUMMARY)

print("\n" + "=" * 95)
print("STRICT WALK-FORWARD BACKTEST COMPLETE")
print("=" * 95)
```

```python
# NORTHGATE AI — STRICT WALK-FORWARD PERFORMANCE & RISK AUDIT

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_ROOT = Path(r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

FEATURES_PATH = PROJECT_ROOT / "data" / "processed" / "features.parquet"
STRICT_PATH = PROJECT_ROOT / "results" / "portfolio" / "strict_walk_forward_daily_returns.csv"
OUTPUT_DIR = PROJECT_ROOT / "results" / "portfolio"

UNIVERSE = [
    "AAPL","MSFT","JPM","XOM","JNJ",
    "PG","NVDA","KO","CAT","HD"]

INITIAL_CAPITAL = 100000.0

# Load strict walk-forward returns

wf = pd.read_csv(
    STRICT_PATH,
    parse_dates=["date"])

wf = (
    wf.sort_values("date")
      .drop_duplicates("date")
      .reset_index(drop=True))

wf_returns = wf.set_index("date")["portfolio_return"].astype(float)

# Load prices / benchmark

df = pd.read_parquet(FEATURES_PATH)

prices = pd.DataFrame(index=df.index)

for ticker in UNIVERSE:
    prices[ticker] = df[("Adj Close", ticker)]

sp500 = df[("Adj Close", "^GSPC")].rename("SP500")

prices = prices.dropna()
sp500 = sp500.loc[prices.index]

stock_returns = prices.pct_change()
market_returns = sp500.pct_change()

# Align strictly to strict strategy dates
stock_returns = stock_returns.reindex(wf_returns.index)
market_returns = market_returns.reindex(wf_returns.index)

# Equal-weight benchmark
equal_weights = np.ones(len(UNIVERSE)) / len(UNIVERSE)
equal_returns = stock_returns[UNIVERSE].dot(equal_weights)

# Wealth curves

wf_wealth = INITIAL_CAPITAL * (1 + wf_returns).cumprod()
equal_wealth = INITIAL_CAPITAL * (1 + equal_returns).cumprod()
sp500_wealth = INITIAL_CAPITAL * (1 + market_returns).cumprod()

# Metrics

def calculate_metrics(returns, wealth, benchmark):

    returns = returns.dropna()
    wealth = wealth.loc[returns.index]

    total_return = wealth.iloc[-1] / INITIAL_CAPITAL - 1

    years = len(returns) / 252

    annualized_return = (
        (1 + total_return) ** (1 / years) - 1)

    volatility = returns.std() * np.sqrt(252)

    sharpe = (
        annualized_return / volatility
        if volatility > 0 else np.nan)

    downside = returns[returns < 0]

    downside_vol = (
        downside.std() * np.sqrt(252)
        if len(downside) > 1 else np.nan)

    sortino = (
        annualized_return / downside_vol
        if pd.notna(downside_vol) and downside_vol > 0
        else np.nan)

    running_max = wealth.cummax()

    drawdown = wealth / running_max - 1

    max_drawdown = drawdown.min()

    aligned = pd.concat(
        [
            returns.rename("portfolio"),
            benchmark.rename("benchmark")
        ],
        axis=1
    ).dropna()

    beta = (
        aligned["portfolio"].cov(aligned["benchmark"])
        / aligned["benchmark"].var())

    return {
        "total_return": total_return,
        "annualized_return": annualized_return,
        "annualized_volatility": volatility,
        "sharpe": sharpe,
        "sortino": sortino,
        "max_drawdown": max_drawdown,
        "beta_vs_sp500": beta,
        "final_value": wealth.iloc[-1]}

# Calculate all metrics

wf_metrics = calculate_metrics(
    wf_returns,
    wf_wealth,
    market_returns)

equal_metrics = calculate_metrics(
    equal_returns,
    equal_wealth,
    market_returns)

sp500_metrics = calculate_metrics(
    market_returns,
    sp500_wealth,
    market_returns)

comparison = pd.DataFrame([
    {
        "portfolio": "Strict Walk-Forward Signal-Aware MPT",
        **wf_metrics
    },
    {
        "portfolio": "Equal Weight",
        **equal_metrics
    },
    {
        "portfolio": "S&P 500",
        **sp500_metrics}])

# Save metrics

metrics_path = (
    OUTPUT_DIR /
    "strict_walk_forward_performance_metrics.csv")

wealth_path = (
    OUTPUT_DIR /
    "strict_walk_forward_wealth_comparison.csv")

chart_path = (
    OUTPUT_DIR /
    "strict_walk_forward_wealth_comparison.png")

comparison.to_csv(metrics_path, index=False)

wealth = pd.DataFrame({
    "date": wf_returns.index,
    "strict_walk_forward_mpt": wf_wealth.values,
    "equal_weight": equal_wealth.values,
    "sp500": sp500_wealth.values})

wealth.to_csv(wealth_path, index=False)
```

```python
# FINAL RECOMMENDATION ENGINE — INTEGRATION AUDIT
# Northgate AI Stock Predictor

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Users\sun\OneDrive\Desktop\Northgate_AI_Stock_Predictor")

RESULTS_DIR = PROJECT_ROOT / "results"

# Required final artifacts

required_artifacts = {
    "Portfolio research signals":
        RESULTS_DIR / "model_comparison" /
        "portfolio_research_signals.csv",

    "Portfolio SVR latest predictions":
        RESULTS_DIR / "model_comparison" /
        "portfolio_svr_latest_predictions.csv",

    "Portfolio risk snapshot":
        RESULTS_DIR / "model_comparison" /
        "portfolio_risk_snapshot.csv",

    "Robust signal-aware MPT weights":
        RESULTS_DIR / "portfolio" /
        "robust_signal_aware_mpt_weights.csv",

    "Robust signal-aware MPT comparison":
        RESULTS_DIR / "portfolio" /
        "robust_signal_aware_mpt_comparison.csv",

    "Walk-forward results":
        RESULTS_DIR / "portfolio" /
        "walk_forward_rebalance_results.csv",

    "Strict walk-forward returns":
        RESULTS_DIR / "portfolio" /
        "strict_walk_forward_daily_returns.csv",

    "Strict walk-forward metrics":
        RESULTS_DIR / "portfolio" /
        "strict_walk_forward_performance_metrics.csv",

    "Strict walk-forward wealth":
        RESULTS_DIR / "portfolio" /
        "strict_walk_forward_wealth_comparison.csv",

    "DL 8-model comparison":
        RESULTS_DIR / "deep_learning" /
        "deep_learning_8_model_comparison.csv",

    "Sentiment ablation":
        RESULTS_DIR / "sentiment" /
        "finbert_sentiment_ablation_comparison.csv",

    "AAPL final recommendation":
        RESULTS_DIR / "model_comparison" /
        "aapl_final_research_recommendation.csv",

    "AAPL recommendation policy audit":
        RESULTS_DIR / "model_comparison" /
        "aapl_final_policy_audit.csv",}

# File existence audit

audit_rows = []

for artifact_name, artifact_path in required_artifacts.items():

    exists = artifact_path.exists()

    size_kb = (
        artifact_path.stat().st_size / 1024
        if exists else np.nan)

    audit_rows.append({
        "artifact": artifact_name,
        "exists": exists,
        "size_kb": size_kb,
        "path": str(artifact_path)})

artifact_audit_df = pd.DataFrame(audit_rows)

# Print artifact audit

print("=" * 75)
print("FINAL RECOMMENDATION ENGINE — ARTIFACT INTEGRATION AUDIT")
print("=" * 75)

display(artifact_audit_df)

missing_artifacts = artifact_audit_df[
    ~artifact_audit_df["exists"]]

print("\nRequired artifacts :", len(required_artifacts))
print("Available          :", artifact_audit_df["exists"].sum())
print("Missing            :", len(missing_artifacts))

if len(missing_artifacts) == 0:
    print("\n✅ ALL REQUIRED INTEGRATION ARTIFACTS ARE PRESENT")
else:
    print("\n⚠️ MISSING ARTIFACTS:")
    display(missing_artifacts)

# Load key portfolio outputs

signals_df = pd.read_csv(
    required_artifacts["Portfolio research signals"])

predictions_df = pd.read_csv(
    required_artifacts["Portfolio SVR latest predictions"])

risk_df = pd.read_csv(
    required_artifacts["Portfolio risk snapshot"])

weights_df = pd.read_csv(
    required_artifacts["Robust signal-aware MPT weights"])

wf_metrics_df = pd.read_csv(
    required_artifacts["Strict walk-forward metrics"])

# Latest portfolio signal audit

signals_df["date"] = pd.to_datetime(
    signals_df["date"])

latest_signal_date = signals_df["date"].max()

latest_signals = (
    signals_df[
        signals_df["date"] == latest_signal_date
    ]
    .copy()
    .sort_values(
        "predicted_5d_return",
        ascending=False))

# Portfolio weight audit

weights_df = weights_df.copy()

# Handle possible naming differences
if "weight" in weights_df.columns:
    weight_column = "weight"
elif "Weight" in weights_df.columns:
    weight_column = "Weight"
else:
    weight_candidates = [
        col for col in weights_df.columns
        if "weight" in col.lower()]

    if len(weight_candidates) == 0:
        raise ValueError(
            "Could not identify portfolio weight column.")

    weight_column = weight_candidates[0]

total_weight = weights_df[weight_column].sum()

negative_weights = (
    weights_df[weight_column] < -1e-10
).sum()

weights_over_20pct = (
    weights_df[weight_column] > 0.200001
).sum()

# Strict walk-forward summary

print("\n" + "=" * 75)
print("LATEST PORTFOLIO SIGNALS")
print("=" * 75)

print(
    f"Latest signal date: "
    f"{latest_signal_date.date()}")

display(latest_signals)

print("\n" + "=" * 75)
print("PORTFOLIO WEIGHT CHECK")
print("=" * 75)

print(f"Total portfolio weight       : {total_weight:.8f}")
print(f"Negative weights             : {negative_weights}")
print(f"Weights above 20%            : {weights_over_20pct}")

display(weights_df)

print("\n" + "=" * 75)
print("STRICT WALK-FORWARD PERFORMANCE")
print("=" * 75)

display(wf_metrics_df)

# Save integration audit

integration_summary = pd.DataFrame({
    "audit_item": [
        "required_artifact_count",
        "available_artifact_count",
        "missing_artifact_count",
        "latest_signal_date",
        "portfolio_total_weight",
        "negative_weight_count",
        "weights_above_20pct"],

    "value": [
        len(required_artifacts),
        int(artifact_audit_df["exists"].sum()),
        int(len(missing_artifacts)),
        latest_signal_date.date(),
        total_weight,
        negative_weights,
        weights_over_20pct]})

artifact_audit_path = (
    RESULTS_DIR /
    "model_comparison" /
    "final_recommendation_artifact_audit.csv")

integration_summary_path = (
    RESULTS_DIR /
    "model_comparison" /
    "final_recommendation_integration_summary.csv")

artifact_audit_df.to_csv(
    artifact_audit_path,
    index=False)

integration_summary.to_csv(
    integration_summary_path,
    index=False)

print("\n" + "=" * 75)
print("✅ RECOMMENDATION INTEGRATION AUDIT SAVED")
print("=" * 75)

print(artifact_audit_path)
print(integration_summary_path)
```
